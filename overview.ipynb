# Ejemplos de Algoritmos Basados en Árboles

Este notebook contiene ejemplos prácticos de cada una de las familias de modelos listadas en la tabla: Single Tree, Bagging, Random Forest, AdaBoost y Gradient Boosting. Utilizaremos la librería `scikit-learn` para todos los ejemplos con un dataset genérico.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score

# Generamos un dataset de clasificación de prueba
X, y = make_classification(n_samples=1000, n_features=20, n_informative=15, n_redundant=5, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("Dataset generado con éxito.")
print(f"Forma de X_train: {X_train.shape}")
print(f"Forma de X_test: {X_test.shape}")
```

## 1. Single Tree
**Ejemplo Industrial:** `sklearn DecisionTree`
**Combina árboles:** —
**Reduce:** Nada

Un árbol de decisión simple que realiza particiones en los datos basadas en las características para llegar a una clasificación.

```python
from sklearn.tree import DecisionTreeClassifier

# Inicializar y entrenar el modelo
tree_clf = DecisionTreeClassifier(random_state=42)
tree_clf.fit(X_train, y_train)

# Predecir y evaluar
y_pred_tree = tree_clf.predict(X_test)
acc_tree = accuracy_score(y_test, y_pred_tree)
print(f"Precisión (Accuracy) del Single Tree: {acc_tree:.4f}")
```

## 2. Bagging
**Ejemplo Industrial:** `BaggingClassifier`
**Combina árboles:** Paralelo + Votación
**Reduce:** Varianza

Entrena múltiples árboles de decisión en subconjuntos aleatorios de los datos (con reemplazo) y combina sus predicciones mediante una votación mayoritaria.

```python
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier

# Inicializar y entrenar usando DecisionTree como estimador base
bagging_clf = BaggingClassifier(
    estimator=DecisionTreeClassifier(random_state=42),
    n_estimators=50, # Entrenamos 50 árboles en paralelo
    random_state=42
)
bagging_clf.fit(X_train, y_train)

# Predecir y evaluar
y_pred_bagging = bagging_clf.predict(X_test)
acc_bagging = accuracy_score(y_test, y_pred_bagging)
print(f"Precisión (Accuracy) de Bagging: {acc_bagging:.4f}")
```

## 3. Random Forest
**Ejemplo Industrial:** `RandomForestClassifier`
**Combina árboles:** Paralelo + Votación + Diversidad
**Reduce:** Varianza (más)

Similar a Bagging, pero al crear cada división dentro de un árbol, solo considera un subconjunto aleatorio de las características. Esto reduce la correlación entre los árboles y disminuye aún más la varianza.

```python
from sklearn.ensemble import RandomForestClassifier

# Inicializar y entrenar el modelo
rf_clf = RandomForestClassifier(n_estimators=100, random_state=42)
rf_clf.fit(X_train, y_train)

# Predecir y evaluar
y_pred_rf = rf_clf.predict(X_test)
acc_rf = accuracy_score(y_test, y_pred_rf)
print(f"Precisión (Accuracy) de Random Forest: {acc_rf:.4f}")
```

## 4. Boosting (AdaBoost)
**Ejemplo Industrial:** `AdaBoostClassifier`
**Combina árboles:** Secuencial + Pesos
**Reduce:** Sesgo (Bias)

Entrena aprendices débiles de forma secuencial. Cada nuevo árbol presta más atención a las instancias que los árboles anteriores clasificaron incorrectamente, aumentando sus pesos.

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier

# Usamos árboles de profundidad 1 (stumps) como estimadores base
ada_clf = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth=1, random_state=42),
    n_estimators=50,
    random_state=42 # <- Le quitamos la línea del algorithm
)
ada_clf.fit(X_train, y_train)

# Predecir y evaluar
y_pred_ada = ada_clf.predict(X_test)
acc_ada = accuracy_score(y_test, y_pred_ada)
print(f"Precisión (Accuracy) de AdaBoost: {acc_ada:.4f}")
print(f"Precisión (Accuracy) de AdaBoost: {acc_ada:.4f}")
```

## 5. Gradient Boosting con XGBoost (Industrial Standard)
**Librería:** `xgboost`
**Por qué usarlo:** Es una implementación avanzada y altamente optimizada del algoritmo de Gradient Boosting. XGBoost incluye regularización (L1 y L2) para evitar el sobreajuste (overfitting) y es significativamente más rápido que el `GradientBoostingClassifier` de sklearn al procesar grandes volúmenes de datos.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score
import xgboost as xgb

# 1. Generamos el dataset de prueba (esto es lo que te faltaba en memoria)
X, y = make_classification(n_samples=1000, n_features=20, n_informative=15, n_redundant=5, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("Datos generados y cargados en memoria...")

# 2. Inicializamos y entrenamos el modelo XGBoost
xgb_clf = xgb.XGBClassifier(
    n_estimators=100,         # Número de árboles
    learning_rate=0.1,        # Tasa de aprendizaje
    max_depth=3,              # Profundidad máxima de cada árbol
    random_state=42,
    eval_metric='logloss'     # Para evitar warnings
)

# Ahora sí, el .fit() ya encuentra X_train y y_train
xgb_clf.fit(X_train, y_train)

# 3. Predecimos y evaluamos
y_pred_xgb = xgb_clf.predict(X_test)
acc_xgb = accuracy_score(y_test, y_pred_xgb)

print(f"Precisión (Accuracy) de XGBoost: {acc_xgb:.4f}")

# Bonus: Ver la importancia de las características
importances = xgb_clf.feature_importances_
print("\nTop 5 variables que el modelo consideró más importantes:")
for i in np.argsort(importances)[-5:][::-1]:
    print(f"Variable {i}: {importances[i]:.4f}")
```
